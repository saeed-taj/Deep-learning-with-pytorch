# Tensor operation on GPU

```python
torch.cuda.is_available()
```

```python
device = torch.device('cuda')
```

```python
import time

size = 10000

cpu_mat_1 = torch.randn(size , size)
cpu_mat_2 = torch.randn(size , size)


cpu_start_time = time.time()
result = torch.matmul(cpu_mat_1 , cpu_mat_2)
cpu_end_time = time.time() - cpu_start_time

print(f"Time on CPU : {cpu_end_time}")


gpu_mat_1 = cpu_mat_1.to('cuda')
gpu_mat_2 = cpu_mat_2.to('cuda')


gpu_start_time = time.time()
gpu_result = torch.matmul(gpu_mat_1 , gpu_mat_2)
torch.cuda.synchronize()
gpu_end_time = time.time() - gpu_start_time

print(f"Time on GPU : {gpu_end_time:.4f}")

print(f"CPU : {gpu_end_time / cpu_end_time}")
print(f"GPU : {cpu_end_time / gpu_end_time}")
```

```python
x = torch.tensor(6.7)
y = torch.tensor(0.0)

w = torch.tensor(1.0)
b = torch.tensor(0.0)
```

```python
def binary_cross_entropy_loss(target , prediction):
  epsilon = 1e-8
  prediction = torch.clamp(prediction , epsilon , 1 - epsilon) # torch.clamp(input, min, max) If model predicts a probability of exactly 0.0 or 1.0 and turns out to be wrong so it should be b/w...

  return -(target * torch.log(prediction) + (1 - target) * torch.log(1 - prediction))
```

## forward pass

```python
z = w * x + b
```

```python
y_pred = torch.sigmoid(z)
```

```python
loss = binary_cross_entropy_loss(y , y_pred)
```

# backward pass

```python
dloss_dy_pred = (y_pred - y) / y_pred*(1 - y_pred)
```

```python
dy_pred_dz = y_pred * (1 - y_pred)
```

```python
dz_dw = x
dz_db = 1
```

```python
dl_dw = dloss_dy_pred * dy_pred_dz * dz_dw
dl_db = dloss_dy_pred * dy_pred_dz * dz_db
```

```python
print(f"loss wrt weight : {dl_dw}")
print(f"loss wrt bias : {dl_db}")
```
