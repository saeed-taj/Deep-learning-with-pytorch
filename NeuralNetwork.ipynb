# train test split

```python
X_train , x_test , y_train , y_test = train_test_split(df.iloc[:,1:] , df.iloc[:,0] , test_size=0.2)
```

```python
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
x_test = scaler.transform(x_test)
```

### fit(): The scaler calculates the mean and standard deviation of each feature specifically in your training set.transform(): It applies the math formula to actually scale the numbers.The formula for standardization (Z-score) is: \(z = \frac{x - \mu}{\sigma}\)(Where x is your data point, μ is the mean, and σ is the standard deviation).

## . It Prevents Data LeakageIf you use fit_transform() on x_test, the scaler calculates a new mean and standard deviation based on the test set. This is a severe machine learning error called data leakage. Your model inadvertently gains information about the distribution of the test set, leading to overly optimistic and unrealistic performance metrics.


## fit: Computes and stores (mean) and(standard deviation) from X_train.
## transform(X_train): Uses those stored values to scale
## X_train.transform(x_test): Uses those same stored values to scale x_test

### The Training Phase (Building the Scale)Before the exam, you give your students a practice worksheet to study.You look at the worksheet and find that the average score is 50%.You decide to grade on a curve based only on this worksheet.To you, a 50% is now the baseline "Average" (Zero).This is exactly what scaler.fit_transform(X_train) does. It looks at the training data, finds the average, and sets the baseline.Step 2: The Testing Phase (The Blind Exam)Now, it is final exam day. You hand out a brand new Test (x_test).Your students need to take this test to prove they actually understand math, not just memorized the practice sheet.If you use fit_transform() on the test data, it means you change the grading curve based on how they did on the final exam.What if the final exam was incredibly easy, and the new average score is 90%?If you recalculate the curve based on the test, a student who got a 90% is now labeled as just "Average" (Zero).But your machine learning model was trained to think that "Average" means a 50% knowledge level!By changing the curve, a 90% on the test gets treated exactly the same as a 50% on the practice sheet. The numbers lose their actual meaning.The Core ProblemTo grade the final exam fairly, you must use the exact same grading curve (the mean and standard deviation) that you built from the practice worksheet.If you create a brand new curve for the final exam:You are changing the rules of the game halfway through.A score of 80 on the practice sheet and 80 on the final exam will mean completely different things to the model.

```python
encoder = LabelEncoder()
y_train = encoder.fit_transform(y_train)
y_test = encoder.transform(y_test)
```

### fit(): Creates the dictionary / rulebook by looking at your categories. For example: {'Cat': 0, 'Dog': 1, 'Fish': 2}.transform(): Uses that rulebook to actually convert the text strings into numbers.

### If you use fit_transform() on y_test, the encoder throws away the old rulebook and builds a brand new dictionary based purely on the test data alphabet.Imagine this scenario:StepDataWhat fit_transform doesResulting RulebookTraining['Cat', 'Dog']fit learns the categories. transform encodes them.Cat = 0, Dog = 1Testing['Dog', 'Cat']fit learns a new order (alphabetical or order of appearance).

### By using encoder.transform(y_test), you force the encoder to use the original training rulebook (Cat = 0, Dog = 1).Even if the test data comes in a different order like ['Dog', 'Cat'], the encoder looks at its training rulebook and outputs [1, 0]. Now, the numbers mean the exact same thing across both datasets, and your model can evaluate them correctly.

## now conver the numpy arrays to tensors

```python
X_train_tensor = torch.from_numpy(X_train)
x_test_tensor = torch.from_numpy(x_test)
y_train_tensor = torch.from_numpy(y_train)
y_test_tensor = torch.from_numpy(y_test)
```

## Parameters

```python
learning_rate = 0.1
epochs = 25
```

```python
class NeuralNetwork():
  def __init__(self , X):
    self.weights = torch.rand(X.shape[1] , 1 ,dtype=torch.float64, requires_grad=True) # 30 rows 1 column
    self.bias = torch.zeros(1 , dtype=torch.float64 , requires_grad=True)

  def forward_pass(self , X):

    z = torch.matmul(X , self.weights) + self.bias
    y_pred = torch.sigmoid(z)
    return y_pred

  def find_loss(self, y, y_pred):
    # clamp
      epsilon = 1e-7
      y_pred = torch.clamp(y_pred , epsilon , 1 - epsilon )  # clamp(input , min , max)
      loss = -(y_train_tensor * torch.log(y_pred) + 1 - y_train_tensor * torch.log(1 - y_pred)).mean()

      return loss


```

## pipeline

```python
model = NeuralNetwork(X_train_tensor)
model.weights
```

```python
# loop
for epoch in range(epochs):
# forward pass w.x + b
  y_pred = model.forward_pass(X_train_tensor)


# calc loss
  loss = model.find_loss(y_train_tensor , y_pred)


# backward pass
  loss.backward()

# parameters update
  with torch.no_grad():
    model.weights -= learning_rate * model.weights.grad
    model.bias -= learning_rate * model.bias.grad

    # zero grads

    model.weights.grad.zero_()
    model.bias.grad.zero_()

    print(f"Epoch : {epoch + 1} and loss : {loss.item()} ")
```

# Model evaluation

```python
with torch.no_grad():
  y_pred = model.forward_pass(x_test_tensor)

  y_pred = (y_pred > 0.9).float()

  accuracy = (y_pred == y_test_tensor).float().mean()
  print(f"Accuracy : {accuracy.item()}")
```
