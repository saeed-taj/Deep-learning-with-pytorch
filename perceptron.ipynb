```python
df = pd.read_csv("placement_data.csv")
```

```python
sns.scatterplot(x=df['cgpa'],y=df['resume_score'],hue=df['placed'])
```

```python
X = df.iloc[: , 0 : 2] # include col 0 and 1 
y = df.iloc[: , -1]  # select the last col which is ( placed )
```

```python
from sklearn.linear_model import Perceptron
p = Perceptron()
```

It checks the shape of X and sees it has 2 columns (from your 0:2 slice).It creates 2 weights (\(w_1, w_2\)) and 1 bias (\(b\)).It sets their initial values to zero

```python
p.fit(X , y)
```

```python
p.coef_

```

```python
p.intercept_
```

```python
from mlxtend.plotting import plot_decision_regions
```

```python
plot_decision_regions(X.values, y.values , clf=p , legend=2 ) # clf = classifier
```
