```python
# create model class
import torch
import torch.nn as nn

class Model(nn.Module):

  def __init__(self, n_features):
    super().__init__()

    self.linear = nn.Linear(n_features, 1) # linear(no_input , no_output)
    self.sigmoid = nn.Sigmoid()


  def forward(self , n_features):
    out = self.linear(n_features)
    out = self.sigmoid(out)



```

```python
# CREATE DATASET
dataset_features = torch.rand(10 , 5)


# create model
model = Model(dataset_features.shape[1])

# call for forward pass
model(dataset_features)
```

# Neural Network with hidden layers

```python
class Model(nn.Module):

  def __init__(self, n_features):

    super().__init__()

    self.linear1 = nn.Linear(n_features, 3)
    self.relu = nn.ReLU()
    self.linear2 = nn.Linear(3, 1)
    self.sigmoid = nn.Sigmoid()

  def forward(self, X_features):
    out = self.linear1(X_features)
    out = self.relu(out)
    out = self.linear2(out)
    out = self.sigmoid(out)

    return out
```

```python
# create dataset
dataset_features = torch.rand(10, 5)

# create model
model = Model(dataset_features.shape[1])

# call the forward pass
out = model(dataset_features)


model.linear1.weight
```

```python
model.linear2.bias
```

### __init__ : Number of columns only Reserves memory and creates the fixed-size weight matrices.

### forward : Actual data tensor (Rows × Columns)Performs the mathematical operations using the data and the weights.

# sequential containers

```python
class Model(nn.Module):

  def __init__(self, n_features):

    super().__init__()

    self.network = nn.Sequential(
        nn.Linear(n_features, 3),
        nn.ReLU(),
        nn.Linear(3, 1),
        nn.Sigmoid()
    )

  def forward(self, X_features):
    out = self.network(X_features)

    return out
```

```python
# create dataset
dataset_features = torch.rand(10, 5)

# create model
model = Model(dataset_features.shape[1])

# call the forward pass
out = model(dataset_features)
print(out)
```
